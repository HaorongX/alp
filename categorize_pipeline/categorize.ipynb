```python
import pandas as pd

df = pd.read_csv("training_data.csv", encoding="utf-8")
df.columns = ['content', 'target']

df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
le.fit(df['target'])

df['target_encoded'] = le.transform(df['target'])
df.head()
```

```python
from sklearn.model_selection import train_test_split

x = df['content']
y = df['target_encoded']

x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=42)
print(len(x_train), len(y_train))
print(len(x_test), len(y_test))
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# instantiate the vectorizer
vect = CountVectorizer()
vect.fit(x_train)

# from sklearn.feature_extraction.text import TfidfVectorizer

# # Replace CountVectorizer with TfidfVectorizer
# vect = TfidfVectorizer(max_features=5000)
# vect.fit(x_train)

x_train_dtm = vect.transform(x_train)
x_test_dtm = vect.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
model = RandomForestClassifier(n_estimators=100)
model.fit(x_train_dtm, y_train)
y_pred = model.predict(x_test_dtm)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2%}")
```

```python
import pickle

with open('tf_idf.pkl', 'wb') as f:
    pickle.dump((vect, model, le), f)
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Encode
encoder = SentenceTransformer('all-MiniLM-L6-v2')
print("Yes")
x_train_encoded = encoder.encode(x_train.tolist(), show_progress_bar=True)
x_test_encoded = encoder.encode(x_test.tolist(), show_progress_bar=True)

# Train
clf = LogisticRegression(max_iter=10, C=1.0)
clf.fit(x_train_encoded, y_train)

# Evaluate
y_pred = clf.predict(x_test_encoded)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2%}")
```
